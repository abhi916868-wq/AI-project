```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as p
import seaborn as s
```

```python
df=pd.read_csv(r"C:\Users\SPTINT-20\Desktop\survey_lung_cancer_SVM.csv")
```

```python
df.info()
```

```python
df.head(2)
```

```python
x=df.drop(["LUNG_CANCER"],axis=1)
y=df["LUNG_CANCER"].values.reshape(-1,1)
```

```python
corrmat=df.corr(numeric_only=True)
s.heatmap(corrmat,annot=True,fmt='.2f')
p.Figure(figsize=[12,8])
```

```python
from sklearn.model_selection import train_test_split as t 
```

```python
x_train, x_test, y_train, y_test=t(x,y)
```

```python
from sklearn.svm import SVC 
```

```python
sv=SVC()
```

```python
sv.fit(x_train,y_train)
```

```python
y_pred=sv.predict(x_test)
```

```python
from sklearn.metrics import classification_report, accuracy_score
print("The accuracy score is:",accuracy_score(y_test,y_pred)*100)
print()
print("The classification report :\n",classification_report(y_test,y_pred))
```
